```python
#Find Sum and Average of 10 intager
# Take 10 numbers
numbers = []
for i in range(10):
    x = int(input("Enter number: "))
    numbers += [x]
# Find sum
total = 0
for x in numbers:
    total = total + x
# Find average
average = total / 10
print("Sum =", total)
print("Average =", average)
```
