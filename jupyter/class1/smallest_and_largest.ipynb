```python
#find smallest and largest
# Take 7 numbers
numbers = []
for i in range(7):
    x = int(input("Enter number: "))
    numbers += [x]
# Assume first number is smallest and largest
small = numbers[0]
large = numbers[0]
# Check every number
for x in numbers:
    if x < small:
        small = x
    if x > large:
        large = x
# Print result
print("Smallest =", small)
print("Largest =", large)
```
