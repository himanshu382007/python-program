```python
#remove duplicates
# Take numbers
numbers = []
for i in range(10):
    x = int(input("Enter number: "))
    numbers += [x]
# Create new list
new = []
for x in numbers:
    found = 0
    for y in new:
        if x == y:
            found = 1
    if found == 0:
        new += [x]
print(new)
```
