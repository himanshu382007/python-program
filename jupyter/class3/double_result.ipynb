```python
def double_result(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result * 2
    return wrapper
def add(a, b):
    return a + b
add = double_result(add)
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
print("Doubled result:", add(a, b))
```
