```python
def show_info(func):
    def wrapper(*args, **kwargs):
        print("Calling function...")
        result = func(*args, **kwargs)
        print("Function executed.")
        return result
    return wrapper
def square(num):
    return num * num
square = show_info(square)
num = int(input("Enter a number: "))
print("Square:", square(num))
```
