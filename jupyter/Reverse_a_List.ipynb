```python
#Reverse a List
# Take 5 numbers
numbers = []
for i in range(5):
    x = int(input("Enter number: "))
    numbers += [x]
# Create empty list
new = []
# Read list from last to first
for i in range(4, -1, -1):
    new += [numbers[i]]
# Print result
print("Original list:", numbers)
print("Reversed list:", new)
```
